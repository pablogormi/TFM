### Sick (Thyroid Disease)

Para Sick, eliminamos el final de las líneas del archivo `sick.data` eliminando las cadenas que verifiquen el regex `.\|\d+$`. Esto se ha hecho mediante un editor de texto, y el archivo limpio se encuentra en `datasets/sick/sick-cleaned.csv`. Notar que el archivo no tiene cabeceras por lo que se las damos nosotros.

```python
df = pd.read_csv("datasets/sick/sick-cleaned.csv", header=None)

# COlumnas extraídas de sick.names
df.columns = ["age", 
              "sex", 
              "on_thyroxine", 
              "query_on_thyroxine",
              "on_antithyroid_medication", 
              "sick", 
              "pregnant", 
              "thyroid_surgery", 
              "I131_treatment", 
              "query_hypothyroid", 
              "query_hyperthyroid", 
              "lithium", 
              "goitre", 
              "tumor", 
              "hypopituitary", 
              "psych", 
              "TSH_measured", 
              "TSH", 
              "T3_measured", 
              "T3", 
              "TT4_measured", 
              "TT4", 
              "T4U_measured", 
              "T4U", 
              "FTI_measured", 
              "FTI", 
              "TBG_measured", 
              "TBG", 
              "referral_source",
              "label"]
```

```python
df.head()
```

De nuevo hay muchas muestras con valoes inválidos ("?"), los convertimos a NaN.

```python
df = df.replace(to_replace='?',value=np.nan)
```

```python
for col in df.columns:
    print(df[col].value_counts())
```

Hay algunas que son claramente numéricas, como age. Las convertimos:

```python
cols = ['age','FTI','TSH','T3','TT4','T4U']
for i in cols:
    df[i] = pd.to_numeric(df[i])

df.info
```

La mayoría de columnas de `TBG` parecen ser NaN, vemos cuantas hay. Lo mismo para otras columnas cuyos valores parecen ser constantes.

```python
print("TBG", sum(df["TBG"].isna()))

bool_cols = [
    col for col in df.select_dtypes(include=["object", "string"]).columns
    if df[col].dropna().isin(["t", "f"]).all() and df[col].notna().any()
]

for col in bool_cols:
    print(col, sum(df[col]=="f"))
 
```

```python
df[bool_cols] = df[bool_cols].map({"t": True, "f": False}.get).astype("boolean")
```

```python
df = df.drop(["TBG", "TBG_measured"], axis=1)
```

```python
df.head()
```

```python
df.isnull().sum()
```
